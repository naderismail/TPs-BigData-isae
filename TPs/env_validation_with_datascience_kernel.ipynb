```python
required = {
    "pyspark":"3.3.0",
    "kafka":"2.0.2",       # kafka-python module name is 'kafka'
    "pandas":None,
    "pyarrow":None,
    "joblib":None,
    "matplotlib":None,
    "seaborn":None,
    "fastavro":None
}
```

```python
from packaging import version
import importlib, sys

ok = True
for modname, minv in required.items():
    try:
        m = importlib.import_module(modname)
        ver = getattr(m, "__version__", None)
        if minv and ver:
            if version.parse(ver) < version.parse(minv):
                print(f"❌ {modname} version {ver} < required {minv}")
                ok = False
            else:
                print(f"✅ {modname} version {ver} OK")
        else:
            print(f"✅ {modname} imported (version={ver})")
    except Exception as e:
        print(f"❌ {modname} import FAILED: {e}")
        ok = False

print("\nOverall:", "ALL GOOD ✅" if ok else "MISSING/OUTDATED ✅")
```
